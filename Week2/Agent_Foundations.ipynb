# Task 1: Agent Concepts & Mental Model
This section explains the basic concepts behind AI agents and how they differ from traditional chatbots and workflows. It also introduces the ReAct pattern and discusses when using an agent would be unnecessary.


## Agent vs Chatbot vs Workflow

### Chatbot
A chatbot is a system that receives a user's message and generates a response. It usually focuses on conversation and does not independently decide to perform multiple external actions.
### Workflow
A workflow is a predefined sequence of steps. The developer decides the order of operations in advance, and the system follows those steps consistently.
### Agent
An agent is a system that receives a goal and dynamically decides what actions are needed to accomplish it. It can use tools, observe their results, perform multiple steps, maintain state, and change its approach when something goes wrong.
### Main Difference
The main difference is **decision-making**. A chatbot mainly responds, a workflow follows predefined steps, while an agent can decide which step or tool should be used next based on the current situation.


## What Makes a System Agentic?
A system becomes more agentic when it can operate with some degree of autonomy instead of simply producing a direct response.
### 1. Autonomy
The agent can decide what action should be performed next without the developer explicitly specifying every step.
### 2. Tool Use
The agent can call external functions or tools such as calculators, databases, APIs, file readers, or weather services.
### 3. Multi-Step Planning
The agent can break a complex goal into several smaller actions and perform them sequentially.
### 4. Observation
After using a tool, the agent receives the result and uses that information to decide what to do next.
### 5. Self-Correction
If an action fails or produces an unexpected result, the agent can adjust its approach instead of immediately stopping.
### 6. State
The agent can maintain information about the current task, previous actions, tool results, and other relevant information.


## ReAct Pattern

The ReAct pattern describes an agent as an iterative loop:

**Reason → Act → Observe → Repeat**

1. **Reason**  : The model determines what needs to be done.
2. **Act**     : The agent selects and calls an appropriate tool.
3. **Observe** : The agent receives the result from the tool.
4. **Repeat**  :  The model uses the observation to decide whether another action is necessary.
5. **Finish**  :  When the task is complete, the agent provides the final answer.
### Example
Suppose the user asks:
"Which city is warmer, Karachi or Lahore?"
The agent can follow this process:
1. Reason: Weather information is required for both cities.
2. Act: Call the weather tool for Karachi.
3. Observe: Karachi = 34°C.
4. Act: Call the weather tool for Lahore.
5. Observe: Lahore = 31°C.
6. Reason: Compare both temperatures.
7. Finish: Karachi is warmer by 3°C.

The agent repeats the Reason → Act → Observe cycle until it has enough information to complete the user's request.


## ReAct Pseudocode

```text
START

Receive user goal

WHILE task is not complete:

    Reason about the next required action

    IF a tool is required:
        Select a tool
        Generate tool arguments
        Execute the tool
        Observe the tool result
        Continue the loop

    ELSE:
        Generate final answer
        STOP

END
```



## When Is an Agent Overkill?

An agent is overkill when a task can be solved reliably using a simple prompt, a normal Python function, or a fixed workflow. Agents introduce additional complexity, API calls, latency, cost, and possible failure points. Therefore, an agent should be used when dynamic decision-making, tool usage, or multi-step reasoning provides a real benefit.
### Example
For example, if a user asks, **"Convert 10 kilometers into miles,"** using an AI agent would be unnecessary. A simple Python function or calculator can solve the problem directly and more efficiently. However, if the user asks the system to compare distances between multiple cities, search for relevant data, perform calculations, and explain the results, an agent may be useful because the task requires multiple steps and tool usage.
