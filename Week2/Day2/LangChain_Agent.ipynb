## LCEL Pipe  Explanation
LangChain Expression Language (LCEL) uses the pipe (`|`) operator to connect runnable components into a sequence. The output of one component is automatically passed as the input to the next component.
In this notebook,prompt | llm means that the user's question is first formatted by the prompt template and then passed to the Gemini model to generate the response.

## Mapping Raw Python Agent to LangChain

| Day 1 Raw Python Agent | Day 2 LangChain Equivalent |
|---|---|
| Direct Gemini API call | ChatGoogleGenerativeAI |
| Python function/tool | LangChain @tool |
| Manual agent loop | AgentExecutor |
| Manual message history | LangChain message history |
| Manual state handling | LangChain Runnable/agent state |
| Manual prompt construction | ChatPromptTemplate |
| Manual result passing | LCEL pipeline |
