# Task 1 — System Design Pick

## Use Case: AI Client Onboarding & Project Proposal Agent

The proposed system is an AI-powered client onboarding and project proposal agent designed for a freelancing or software business scenario.

The system accepts a client's project request, validates the input, analyzes the client's requirements, researches relevant information when needed, generates a project proposal, performs a quality and safety check, and sends the final proposal for human approval before it is considered ready for delivery.

The main goal is to reduce the time required to understand client requirements and prepare professional project proposals while keeping a human involved before any consequential output is finalized.

## Business Objective

The system aims to:
1. Understand and structure incoming client requirements.
2. Identify important project requirements, constraints, and missing information.
3. Use external information when research is required.
4. Generate a clear and professional project proposal.
5. Validate the generated proposal for quality and safety.
6. Require human approval before finalizing the proposal.
7. Provide a reliable API-based interface for future integration.

## Proposed Workflow
1. Receive client request.
2. Validate the input.
3. Analyze and structure project requirements.
4. Perform research using an external data source when required.
5. Generate a project proposal.
6. Perform quality and safety checks.
7. Ask for human approval.
8. Return the approved final proposal.

## Framework Choice
LangGraph is selected as the primary framework because this system requires a controlled, stateful workflow with conditional routing, validation, error handling, and a human approval checkpoint. LangGraph allows each stage to be represented as a node while maintaining shared state throughout the workflow. A human-in-the-loop checkpoint is important because the final proposal should be reviewed before it is delivered to a client. CrewAI could be useful for role-based collaboration, but LangGraph is a better fit here because workflow control and explicit state transitions are more important than independent agent roles.

## System Components
### 1. Input Validation Node
Receives the client's request and checks whether the input contains sufficient and valid information.

### 2. Requirement Analysis Node
Analyzes the client's request and extracts project requirements, goals, constraints, technologies, and important missing information.

### 3. Research Node
Uses an external web search/data source when additional information is required for the proposal.

### 4. Proposal Generation Node
Uses the validated requirements and research findings to generate a structured project proposal.

### 5. Quality & Safety Check Node
Reviews the generated proposal for completeness, relevance, unsupported claims, unsafe content, and formatting problems.

### 6. Human Approval Checkpoint
The proposal is presented to a human reviewer before finalization. The reviewer can approve the proposal or send it back for revision.

### 7. Final Output Node
Returns the approved proposal in a structured format.
