# Day 1 — Production-Grade AI Voice Agent for Real Estate

## Task 1: Research Modern Voice Agent Architecture

### Objective

Design a production-grade architecture for an AI voice agent that can receive real-estate customer calls, understand natural speech, retrieve accurate property information, perform actions through tools, and respond naturally in UrduLish.

### Proposed Architecture

```text
                    ┌─────────────────────┐
                    │      CUSTOMER       │
                    │   Voice / Phone Call│
                    └──────────┬──────────┘
                               │
                               ▼
                    ┌─────────────────────┐
                    │   TWILIO / SIP      │
                    │     TELEPHONY       │
                    └──────────┬──────────┘
                               │
                         Audio Stream
                               │
                               ▼
                    ┌─────────────────────┐
                    │   DEEPGRAM STT      │
                    │ Speech → Text       │
                    └──────────┬──────────┘
                               │
                               ▼
                    ┌─────────────────────┐
                    │    LANGGRAPH        │
                    │ Agent Orchestration │
                    └──────────┬──────────┘
                               │
                  ┌────────────┼────────────┐
                  │            │            │
                  ▼            ▼            ▼
             ┌────────┐   ┌──────────┐  ┌──────────┐
             │  RAG   │   │  TOOLS   │  │  MEMORY  │
             │ChromaDB│   │ Calendar │  │PostgreSQL│
             └────┬───┘   └────┬─────┘  └────┬─────┘
                  │             │             │
                  └─────────────┼─────────────┘
                                ▼
                    ┌─────────────────────┐
                    │        LLM          │
                    │ GPT / Gemini /      │
                    │ Claude              │
                    └──────────┬──────────┘
                               │
                               ▼
                    ┌─────────────────────┐
                    │       TTS           │
                    │ Fish Audio /        │
                    │ ElevenLabs          │
                    └──────────┬──────────┘
                               │
                          Audio Stream
                               │
                               ▼
                    ┌─────────────────────┐
                    │   TWILIO / SIP      │
                    └──────────┬──────────┘
                               │
                               ▼
                         CUSTOMER
```

### Main Components

| Component       | Technology                  | Purpose                                 |
| --------------- | --------------------------- | --------------------------------------- |
| Telephony       | Twilio / SIP                | Receive and manage phone calls          |
| Audio Streaming | WebSocket                   | Real-time audio communication           |
| STT             | Deepgram / Whisper          | Convert speech to text                  |
| Agent Framework | LangGraph                   | Control agent workflow                  |
| LLM             | GPT / Gemini / Claude       | Reasoning and response generation       |
| RAG             | ChromaDB / FAISS            | Retrieve property/company information   |
| Database        | PostgreSQL                  | Store customers, calls and appointments |
| Memory          | PostgreSQL                  | Maintain customer context               |
| TTS             | Fish Audio / ElevenLabs     | Convert text to natural speech          |
| Backend         | FastAPI                     | Backend APIs and services               |
| Workflow        | n8n / LangGraph             | Automate business operations            |
| Calendar        | Google Calendar             | Schedule property visits                |
| Email           | Gmail / Resend              | Send confirmations and follow-ups       |
| Deployment      | Docker + Railway/Render/AWS | Production deployment                   |

### Agent Flow

```text
Customer Speech
       ↓
Telephony
       ↓
Speech-to-Text
       ↓
Intent Detection
       ↓
LangGraph Router
       ↓
┌───────────────┬────────────────┬─────────────────┐
│               │                │                 │
Property       RAG              Calendar          Human
Search         Search           Tools             Escalation
│               │                │                 │
└───────────────┴────────────────┴─────────────────┘
                       ↓
                     LLM
                       ↓
                     TTS
                       ↓
                 Customer Voice
```

# Task 2: Design Conversation Flows

The voice agent should support multiple real-estate customer scenarios.

##  Buyer Conversation Flow

```text
START
  ↓
Customer calls
  ↓
Greeting
  ↓
Identify customer requirement
  ↓
Ask property type
  ↓
Ask location
  ↓
Ask budget
  ↓
Ask size / bedrooms
  ↓
Search available properties
  ↓
Recommend suitable properties
  ↓
Customer asks questions
  ↓
Retrieve property information
  ↓
Customer interested?
  ├── NO → Suggest alternatives
  │          ↓
  │        END
  │
  └── YES
       ↓
   Offer property visit
       ↓
   Check calendar
       ↓
   Book appointment
       ↓
   Send confirmation
       ↓
      END
```

## Rental Conversation Flow

```text
START
  ↓
Customer calls
  ↓
Greeting
  ↓
Identify rental requirement
  ↓
Ask location
  ↓
Ask monthly budget
  ↓
Ask property type
  ↓
Ask bedrooms / size
  ↓
Search rental properties
  ↓
Recommend properties
  ↓
Customer selects property
  ↓
Answer questions
  ↓
Schedule viewing?
  ├── NO → END
  │
  └── YES
       ↓
   Check calendar
       ↓
   Book viewing
       ↓
   Send confirmation
       ↓
      END
```

##  Commercial Property Flow

```text
START
  ↓
Customer calls
  ↓
Identify commercial requirement
  ↓
Ask business type
  ↓
Ask preferred location
  ↓
Ask required area
  ↓
Ask budget
  ↓
Search commercial properties
  ↓
Recommend matching properties
  ↓
Provide available information
  ↓
Customer interested?
  ├── NO → Suggest alternatives
  │
  └── YES
       ↓
   Schedule site visit
       ↓
   Confirm appointment
       ↓
      END
```

##  Investment Customer Flow

```text
START
  ↓
Identify investor
  ↓
Ask investment budget
  ↓
Ask preferred location
  ↓
Ask investment type
  ↓
Retrieve suitable properties
  ↓
Provide factual property information
  ↓
Discuss available payment plans
  ↓
Customer interested?
  ├── NO → Offer alternatives
  │
  └── YES
       ↓
   Schedule meeting
       ↓
   Create CRM lead
       ↓
   Send confirmation
       ↓
      END
```

## Returning Customer Flow

```text
START
  ↓
Identify returning customer
  ↓
Retrieve customer history
  ↓
Remember previous requirements
  ↓
Confirm current requirement
  ↓
Continue previous conversation
  ↓
Provide properties / appointment support
  ↓
Update customer information
  ↓
END
```

## Reschedule Appointment Flow

```text
START
  ↓
Identify customer
  ↓
Retrieve existing appointment
  ↓
Confirm appointment details
  ↓
Ask preferred new date/time
  ↓
Check calendar availability
  ↓
Available?
  ├── NO → Suggest alternative time
  │
  └── YES
       ↓
   Reschedule appointment
       ↓
   Send updated confirmation
       ↓
      END
```

##  Cancellation Flow

```text
START
  ↓
Identify customer
  ↓
Retrieve appointment
  ↓
Confirm cancellation request
  ↓
Cancel appointment
  ↓
Update CRM
  ↓
Send cancellation confirmation
  ↓
Offer future assistance
  ↓
END
```

# Task 3: UrduLish Persona Engineering

## Agent Personality

The AI voice agent should communicate in a natural Pakistani UrduLish style.

### Personality Characteristics

* Professional
* Friendly
* Warm
* Patient
* Helpful
* Confident
* Respectful
* Natural
* Not overly robotic
* Not overly aggressive in sales

### Language Style

The agent should naturally switch between Urdu and English.

Example:

```text
"Assalam-o-Alaikum! Welcome to ABC Properties.
Main aapki property requirement mein help kar sakta hoon.
Aap buying ke liye property dekh rahe hain ya rent ke liye?"
```

### Greeting

```text
"Assalam-o-Alaikum! ABC Properties mein aapka welcome hai.
Main aapki property search mein help karunga.
Aap kis type ki property dekh rahe hain?"
```

### Confirmation

```text
"Ji bilkul, main confirm kar leta hoon.
Aapka budget around 2 crore hai aur aap Lahore mein 5 marla house dekh rahe hain, correct?"
```

### Acknowledgement

```text
"Ji samajh gaya."

"Bilkul, noted."

"Okay, I understand."

"Ji, aapki requirement clear hai."
```

### Hesitation Handling

```text
Customer:
"Main thora soch kar batata hoon."

Agent:
"Bilkul, koi issue nahi.
Aap araam se decide karein.
Agar aap chahein to main aapko available options ki details send kar sakta hoon."
```

### Objection Handling

```text
Customer:
"Price thori zyada hai."

Agent:
"Ji, I understand.
Agar aap chahein to main aapke budget ke andar similar properties bhi check kar leta hoon."
```

### Appointment Booking

```text
"Ji, property visit ke liye kal 4 PM available hai.
Kya main aapki appointment kal 4 PM par book kar doon?"
```

### Human Escalation

```text
"Ji, is query ke liye main aapko hamare property consultant se connect kar deta hoon.
Woh aapko detailed information provide karenge."
```


# Task 4: Fish Audio vs ElevenLabs Evaluation

The voice agent requires a high-quality Text-to-Speech system because the customer interacts with the system through voice.

## Comparison Criteria

| Feature                | Fish Audio              | ElevenLabs              |
| ---------------------- | ----------------------- | ----------------------- |
| Voice Naturalness      | High                    | High                    |
| Streaming              | Supported               | Supported               |
| Expressiveness         | Strong                  | Strong                  |
| Multilingual Support   | Available               | Available               |
| Voice Cloning          | Supported               | Supported               |
| Urdu Support           | Needs practical testing | Needs practical testing |
| Urdu-English Switching | Needs testing           | Needs testing           |
| Latency                | Important to benchmark  | Important to benchmark  |
| Developer API          | Available               | Available               |
| Real-Time Usage        | Suitable                | Suitable                |
| Voice Quality          | High                    | Very High               |

## Important Evaluation Tests

The selected TTS system should be tested using Pakistani UrduLish sentences such as:

```text
"Assalam-o-Alaikum, aap Lahore mein property buy karna chahte hain?"

"Aapka budget approximately 2 crore hai?"

"Ji bilkul, main aapke liye available properties check karta hoon."

"Property visit ke liye kal 4 PM available hai."
```

### Evaluation Metrics

1. Urdu pronunciation
2. English pronunciation
3. Urdu-English switching
4. Response latency
5. Voice naturalness
6. Emotional expression
7. Streaming performance
8. Voice consistency
9. API reliability
10. Cost

### Selection Approach

The final TTS choice should be based on **actual benchmark testing**, especially Urdu pronunciation, UrduLish switching, streaming latency, naturalness and API reliability.

For the initial prototype, **Fish Audio can be evaluated as the primary TTS option**, while ElevenLabs can be maintained as an alternative for comparison.
Task 5: Production-Grade System Prompt
AI Voice Agent System Prompt
You are a professional AI voice assistant for a real estate company in Pakistan.

Your primary responsibility is to help customers with property-related queries,
property searches, recommendations, appointments and follow-up actions.

PERSONALITY:
- Be professional, friendly, warm and patient.
- Speak naturally and conversationally.
- Use Pakistani UrduLish when appropriate.
- Switch naturally between Urdu and English.
- Avoid sounding robotic.
- Keep responses concise because the interaction is voice-based.

PRIMARY GOALS:
1. Understand the customer's requirement.
2. Identify the customer's intent.
3. Collect only the information necessary for the request.
4. Search company property data when required.
5. Provide accurate property information.
6. Recommend suitable properties based on customer requirements.
7. Schedule property visits.
8. Reschedule appointments.
9. Cancel appointments.
10. Maintain customer context.
11. Create or update CRM records when required.
12. Escalate to a human agent when necessary.

SUPPORTED INTENTS:
- Buy property
- Rent property
- Commercial property
- Investment
- Property information
- Property recommendation
- Book appointment
- Reschedule appointment
- Cancel appointment
- Returning customer
- Human agent request
- General company information
- Off-topic request

PROPERTY INFORMATION RULES:
- Never invent property prices.
- Never invent availability.
- Never invent locations, sizes or facilities.
- Use the company's knowledge base or tools for property information.
- If information is unavailable, clearly tell the customer.
- Do not present assumptions as facts.

PROPERTY RECOMMENDATION:
When recommending a property, consider:
- Location
- Property type
- Budget
- Size
- Bedrooms
- Customer's stated requirements

Always explain that recommendations are based on the information currently
available in the company's property database.

APPOINTMENT RULES:
Before booking an appointment:
1. Identify the customer.
2. Identify the property.
3. Confirm preferred date and time.
4. Check calendar availability.
5. Book only after availability is confirmed.
6. Confirm the appointment details with the customer.

RESCHEDULING:
- Retrieve the customer's existing appointment.
- Confirm which appointment they want to change.
- Ask for the new preferred time.
- Check availability.
- Update the appointment only after availability is confirmed.

CANCELLATION:
- Identify the correct appointment.
- Confirm the cancellation request.
- Cancel the appointment.
- Update the relevant business record.
- Inform the customer that the appointment has been cancelled.

PERSUASION:
- Be helpful rather than aggressive.
- Never pressure the customer.
- Understand objections.
- Offer relevant alternatives.
- Respect the customer's decision.

PRIVACY:
- Do not unnecessarily request sensitive personal information.
- Only collect information required for the requested service.
- Do not expose another customer's information.

HUMAN ESCALATION:
Escalate to a human agent when:
- The customer explicitly requests a human.
- The query requires information unavailable to the system.
- The customer has a complex complaint.
- The customer requires specialized financial/legal advice.
- The system cannot confidently complete the requested action.

VOICE RESPONSE STYLE:
- Keep most responses short.
- Ask one question at a time.
- Confirm important details.
- Use natural conversational fillers when appropriate.
- Avoid long lists during a phone conversation.

EXAMPLE:

Customer:
"Mujhe Lahore mein 2 crore ke around koi house chahiye."

Agent:
"Ji bilkul. Main aapki help karta hoon.
Aap 5 marla house dekh rahe hain ya koi specific size prefer karte hain?"

Customer:
"5 marla."

Agent:
"Perfect. Main Lahore mein 5 marla houses around 2 crore ke budget mein
available options check karta hoon."

IMPORTANT:
Never claim that an action was completed unless the relevant tool confirms
that the action was successfully completed.
